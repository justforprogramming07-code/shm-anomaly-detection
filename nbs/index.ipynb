# Problem

Over time, bridges undergo changes. As the material used ages, it loses strength and the bridge's structural integrity keeps degrading. If there are no sensors fitted, manual visual inspection is the way to go, which does not even guarantee if the internal damage will be flagged or not. If there are structural health monitoring sensors fitted, then the work is done manually, but if we use data-driven methods, the chance of missing something becomes minimal.

# What I Built

A machine learning pipeline that detects the anomalies after separating the environmental variables because the reading is also affected by the environment and there are a lot of correlation we need to separate the affects from the actual reading of just the bridge frequencies 

# How does it work?


There are two ways it is done, one is through Isolation Forest and the second using Autoencoder.

Bridge vibration frequencies are recorded continuously. First, we use a linear regression model to separate the pure bridge vibration frequency from environmental factors and then pass the pure vibration frequency data into our Isolation Forest model which, speaking at a higher level, isolates points by randomly cutting the data space. The anomalies get isolated quickly and they are flagged. While in the case of our other pipeline, we first separate the environmental parameters using the Random Forest method and then pass the pure frequency data into our Autoencoder which, again speaking on a higher level, first learns what normal looks like and then flags anything that does not look like normal as anomalies.

# What did it find?

Both models successfully flagged the May 2019 retrofitting event on the KW51 railway bridge a real structural intervention without being told it happened. The anomaly shows up as a permanent shift in the bridge's vibration pattern after that date.
